# **3. Tokenization**
- 텍스트를 모델이 처리할 수 있는 단위인 Token으로 나누는 과정
- [토큰 계산](https://platform.openai.com/tokenizer)

### 토큰
- 토큰은 모델이 처리하는 텍스트의 작은 단위
- 하나의 토큰은 단어, 부분적인 단어, 공백이나 구두점일 수 있음
- 토큰은 반드시 단어 하나를 의미하지는 않음

### Vocabualry
- Vocabualry는 Tokenizer가 사용하는 Token들의 집합 
- 각 토큰에는 고유한 정수 ID가 할당됨
- 정수 ID 자체는 언어적인 의미가 전혀 없음


> 원핫인코딩: 범주형 데이터를 컴퓨터(모델)가 계산할 수 있도록 0과 1로 이루어진 벡터(수치형 데이터)로 변환해 주는 전처리 작업


> Tokenizer
> 1. 모델(학습, 추론 등)에 넣기 전에 토큰화 시켜주는 프로그램
> 2. 텍스트 데이터를 컴퓨터(모델)가 연산할 수 있도록 작은 단위(Token)로 쪼개고, 이를 고유한 숫자(ID)로 변환(Encoding)해 주는 전처리 프로그램

### 허깅 페이스
허깅 페이스(Hugging Face)는 자연어 처리뿐 아니라 이미지, 음성, 영상, 멀티 모달 모델을 공유하고 사용하는데 필요한 여러 도구와 서비스를 제공하는 AI 생태계이자 회사 이름

````python
import torch
````

````python
if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
elif torch.xpu.is_available():
    DEVICE = torch.device('xpu')
else:
    DEVICE = torch.device('cpu')

print(DEVICE)
````

````python
text = " 나는 자연어를 공부한다"   # 공부한다. 이렇게 되면 공부한다. 이게 1로 빠짐
tokens = text.split()
# tokens
vocab = {"<PAD>": 0,"<UNK>": 1, "나는": 2, "자연어를": 3, "공부한다":4}  # <- 보통 이런 형식으로 들어가있음. 그리고 이렇게 만드는 거임????. 패딩을 넣는 방식에 0을 넣음
token_ids = [vocab.get(token, vocab['<UNK>']) for token in tokens]
# token_ids

input_ids = torch.tensor(token_ids, dtype=torch.long)

print("원문: ", text)
print("Token: ", tokens)
print("Token ID: ", token_ids)
print("PyTorch Tensor: ", input_ids)
print("dtype: ", input_ids.dtype)
````

````python
from transformers import AutoTokenizer  # pip install transformers
````

````python
# 허깅 페이스 허브의 "klue/bert-base" Repository에서 Tokenizer 설정과 Voca를 불러옴
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")  # klue는 한국어에 관려된 데이터(말뭉치) bert-base모델을 사용함
````

````python
text = "자연어 처리를 공부합니다."
tokens = tokenizer.tokenize(text)
print("Token", tokens)  # 한국어는 조사 때문에 ##어, ##를 뜻이 달라질 수 있기 때문에 분리함
# 이건 토크나이저에 따라 다를 수 있음

token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("token ID: ", token_ids)
````

````python
print("Vocabulary 크기: ", tokenizer.vocab_size)
print("[PAD] ID: ", tokenizer.pad_token_type_id)
print("[UNK] ID: ", tokenizer.unk_token_id)
print("[CLS] ID: ", tokenizer.cls_token_id)
print("[SEP] ID: ", tokenizer.sep_token_id)
````

### OOV(Out Of Vocabulary)
- OOV는 Voca에 존재하지 않는 Token이 입력되는 문제
- 과거의 단어 단위 Voca에서는 처음 보는 단어를 \<UNK>와 같은 Unknown Token으로 처리하는 경우가 많았음
- 현대 NLP에서는 Subword Tokenization을 이용해 이 문제를 크게 줄임

````python
vocab = {"<PAD>": 0,"<UNK>": 1, "나는": 2, "파이토치를": 3, "공부한다":4}
tokens = {"나는", "트랜스포머를", "공부한다"}
token_ids = [vocab.get(token, vocab['<UNK>']) for token in tokens]
print(token_ids) # 이거 출력값은 큰 의미가 없음. for문으로 돌려서 그럼. 이걸 for문을 써서 돌리는 건 아니기 때문
````

### Token ID
Tokenization이 끝나면 각각의 Token을 Vocabulary에 정의된 정수 ID로 변환

```
"오늘 날씨가 좋다" > Tokenization > [Token1, Token2, Token3 ...] > Token ID > [102, 235, 303, ...]
```
